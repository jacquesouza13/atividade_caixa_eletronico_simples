```python
# sistema de caixa eletrônico
def depositar(saldo):
  # fazer o código
  # perguntar qual valor quer depositar
  valor = float(input("Qual o valor: R$ "))

  # fazer o algoritmo do deposito (saldo + deposito)
  novo_saldo = saldo + valor
  # informar o novo saldo
  return f"seu novo saldo é {novo_saldo}"



def sacar(saldo):
  # fazer o codigo
  # informar ao usuario o saldo atual
  print(f"seu saldo atual é {saldo}")


  # perguntar qual valor quer sacar
  valor = float(input("Qual o valor: R$ "))
  if valor <= saldo:
    # fazer o algoritmo do saque (saldo - saque)
    novo_saldo = saldo - valor
    # informar o novo saldo
    return f"seu novo saldo é {novo_saldo}"
  else:
    return f"saldo insuficiente"


  # opcional - se o saldo for menor que o saque,
  # nao permitir

```

```python
print("===== Bem vindo ao GEN-Caixa =====")
saldo = float(input("Informe seu saldo"))

print("Escolha uma das opções")
print("1 - ver saldo")
print("2 - depositar")
print("3 - sacar")
print("qualquer outra opção - sair")



resposta = int(input("Digite a opção: "))
if resposta == 1:
  print(f"seu saldo é: {saldo} ")
elif resposta == 2:
  print(depositar(saldo))
elif resposta == 3:
  print(sacar(saldo))
else:
  print("opção invalida")






```
